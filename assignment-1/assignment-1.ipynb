Σύμφωνα με το κείμενο, οι ισοδύναμες συναρτήσεις στάθμισης μοιάζουν πολύ με τις συναρτήσεις πυκνότητας πιθανότητας Gauss όταν a = 0.4. ##### **β) Τι ορίζουμε ως εντροπία; Υπολογίστε τη μέγιστη τιμή εντροπίας που μπορεί να έχει μια grayscale εικόνα;**

Η εντροπία είναι ένα μέτρο της αβεβαιότητας ή του έκτακτου μέσου που απαιτείται για να περιγράψει κάποιο σύστημα. Στο πλαίσιο της εικόνας, η εντροπία μπορεί να αναφέρεται στον βαθμό αβεβαιότητας σχετικά με τα επίπεδα γκρι της εικόνας ή την πιθανότητα εμφάνισης ενός συγκεκριμένου επίπεδου γκρι.

Για μια grayscale εικόνα, όπου κάθε pixel έχει ένα επίπεδο γκρι από 0 έως 255, η μέγιστη εντροπία επιτυγχάνεται όταν κάθε επίπεδο γκρι είναι εξίσου πιθανό να εμφανιστεί. Άρα, η μέγιστη εντροπία `H_max` για μια grayscale εικόνα υπολογίζεται ως:

$H_{\text{max}} = - \sum_{i=0}^{255} f(i) \log_2 f(i)$

όπου `f(i)` είναι η πιθανότητα εμφάνισης κάθε επίπεδου γκρι. Στην περίπτωση της μέγιστης εντροπίας, όλα τα επίπεδα γκρι θα έχουν την ίδια πιθανότητα εμφάνισης, άρα:

$f(i) = \frac{1}{256}$


Έτσι, η μέγιστη εντροπία `H_max` για μια grayscale εικόνα είναι:

$H_{\text{max}} = - \sum_{i=0}^{255} \left(\frac{1}{256}\right) \log_2 \left(\frac{1}{256}\right)$


Υπολογίζοντας αυτό το άθροισμα, βρίσκουμε τη μέγιστη εντροπία που μπορεί να έχει μια grayscale εικόνα.

$H_{\text{max}} = 8$

### Α. Υλοποίηση Αλγορίθμου

```python
import numpy as np
import cv2
from scipy.signal import convolve2d
from skimage.transform import rescale
from skimage import color, data
```

#### **α) Υλοποιήστε την συνάρτηση: h = GKernel(a)**
#### Η συνάρτηση δημιουργεί ένα Generating Kernel, εξαρτώμενο από την παράμετρο ‘a’, (δείτε σελίδα 533 του άρθρου)

```python
def GKernel(a):
    w = np.array([1/4 - a/2, 1/4, a, 1/4, 1/4 - a/2])
    return np.outer(w, w)
```

Aυτή η συνάρτηση δημιουργεί έναν Gaussian πυρήνα βασισμένο στην παράμετρο a. Αυτός ο πυρήνας μπορεί να χρησιμοποιηθεί για λειτουργίες όπως θόλωση ή εξομάλυνση εικόνων. Η παράμετρος a ελέγχει το σχήμα του πυρήνα.

#### **β) Υλοποιήστε την συνάρτηση: I_out = GREDUCE(I, h), σύμφωνα με την εξίσωση (1) (σελίδα 533 του άρθρου)**

```python
def GREDUCE(I, h):
    I_out = cv2.filter2D(I, -1, h)
    I_out = cv2.resize(I_out, (I_out.shape[1]//2, I_out.shape[0]//2), interpolation=cv2.INTER_AREA)
    return I_out
```

Η συνάρτηση μειώνει το μέγεθος μιας εικόνας εφαρμόζοντας Gaussian εξομάλυνση και στη συνέχεια μειώνοντας τη δειγματοληψία της.

#### **γ) Υλοποιήστε την συνάρτηση: G = GPyramid(I, a, depth)**

```python
def GPyramid(I, a, depth):
    pyramid = [I]
    for _ in range(depth):
        h = GKernel(a)
        I = GREDUCE(I, h)
        pyramid.append(I)
    return pyramid
```

Η συνάρτηση <b>GPyramid</b>, δημιουργεί τη $N$-επιπέδων Gaussian πυραμίδα μιας εικόνας $I$ ως:

\begin{equation}
  \left\{\begin{array}{rcl}
    G^0 & = & I \\
    G^{n+1} & = & \text{REDUCE}(G^n);~n\in\{0,1,\cdots,N-1\}
  \end{array}\right.
\end{equation}

#### **δ) Υλοποιήστε την συνάρτηση: I_out = GEXPAND (I, h)**

```python
def GEXPAND(I, h):
    x = cv2.resize(I, (I.shape[1]*2, I.shape[0]*2), interpolation=cv2.INTER_AREA)
    I_out = cv2.filter2D(x,-1,h)
    return x
```

Η συνάρτηση GEXPAND(I, h) εκτελεί το αντίστροφο της GREDUCE. Εάν το GREDUCE μειώνει το μέγεθος μιας εικόνας εφαρμόζοντας Gaussian εξομάλυνση και μειώνοντας τη δειγματοληψία της, το GEXPAND κάνει το αντίθετο, δηλ., αναβαθμίζει την εικόνα και στη συνέχεια εφαρμόζει Gaussian εξομάλυνση.

#### **ε) Υλοποιήστε την συνάρτηση: L = LPyramid(I, a, depth)**

```python
def L_Pyramid(I,a,depth):
    h = GKernel(a)
    G = GPyramid(I, a, depth)
    L = [G[depth-1]]
    for n in range(depth-1):
        interpolation = GEXPAND(G[depth-n-1],h)
        difference = G[depth-n-2] - interpolation + 128
        L.append(difference)
    L.reverse()
    return L
```

Η συνάρτηση <b>L_Pyramid</b>, η οποία χρησιμοποιώντας τη Gaussian Pyramid και τη συνάρτηση EXPAND, δημιουργεί την Laplacian Pyramid μιας εικόνας εισόδου Ι ως:
\begin{equation}
  \left\{\begin{array}{rcl}
    L^{N-1} & = & G^{N-1} \text{ /* the top level of the Gaussian pyramid */}\\
    L^n     & = & G^n - \text{EXPAND}(G^{n+1});~n\in\{N-2,N-3,\cdots,0\}
\end{array}\right.
\end{equation}

#### **στ) Υλοποιήστε την συνάρτηση: I_out = L_Pyramid_Decode(L,a)**

```python
def L_Pyramid_Decode(L,a):
    N = len(L)
    R = L[N-1]
    h = GKernel(a)
    for n in range(N-1):
        R = GEXPAND(R,h)
        R = R + L[N-n-2] - 128
    return R
```

Η συνάρτηση επιστρέφει την αποκωδικοποιημένη εικόνα I_out, λαμβάνοντας ως είσοδο την Laplacian πυραμίδα L και την παράμετρο 'a' που χρησιμοποιήθηκε
για τη δημιουργία της.

#### **ζ) Υλοποιήστε: L_Quantization, σύμφωνα με την εξίσωση (5) (σελ. άρθρου 538)**

```python
def L_Quantization(Im,bins):
    q = np.linspace(0,255,num = bins)
    roll = np.roll(q,-1)
    roll[-1] = 255
    Q = (np.digitize(Im,q) + np.digitize(Im,roll)) //2
    return Q.astype(np.uint8)
```

Αυτή η συνάρτηση εκτελεί κβαντισμό σε μια εικόνα εισόδου Im χρησιμοποιώντας έναν καθορισμένο αριθμό bin. Δημιουργεί ένα σύνολο επιπέδων κβαντοποίησης q με βάση τον αριθμό των κάδων. Υπολογίζει ένα ρολό κυλιόμενου πίνακα που είναι μια μετατοπισμένη έκδοση του q. Υπολογίζει την κβαντισμένη εικόνα Q ψηφιοποιώντας την εικόνα εισόδου Im χρησιμοποιώντας τα επίπεδα κβαντισμού και υπολογίζοντας τον μέσο όρο των αποτελεσμάτων δύο πράξεων ψηφιοποίησης. #### **δ) Υπολογίστε τη εντροπία και παρουσιάστε τα αντίστοιχα διαγράμματα για τα διαφορετικά ‘a’, και ‘depth’ , για κάθε εικόνα και σχολιάστε επαρκώς.**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as st

def histogram(LP, D):
    sum_hist = np.zeros([256,1], np.float64)
    for i in range(D-1):
        hist = cv2.calcHist([LP[i]], [0], None, [256], [0, 256])
        sum_hist += hist
    return sum_hist/sum_hist.max()

def entropy(hist):
    hist = np.squeeze(hist)
    hist = hist[hist != 0]
    return -np.sum(hist * np.log2(hist))

def pyramid_entropy(image, a, depth):
    entropies = []
    for _ in range(3):
        pyramid = L_Pyramid(image, a, depth)
        hist = histogram(pyramid, depth)
        entropy_val = entropy(hist)
        entropies.append(entropy_val)
    return np.mean(entropies)

def plot_entropy_vs_a(image, depths, a_values, image_name):
    entropies = []
    for a in a_values:
        entropy_vals = []
        for depth in depths:
            entropy_val = pyramid_entropy(image, a, depth)
            entropy_vals.append(entropy_val)
        entropies.append(entropy_vals)

    plt.figure(figsize=(10, 6))
    for i, entropy_vals in enumerate(entropies):
        plt.plot(depths, entropy_vals, label=f'a={a_values[i]}')
    plt.xlabel('Depth')
    plt.ylabel('Entropy')
    plt.title(f'Entropy vs Depth for Different Values of "a" ({image_name})')
    plt.legend()
    plt.show()

def plot_entropy_vs_a_and_depth(image, a_values, depths, image_name):
    plt.figure(figsize=(12, 8))
    for depth in depths:
        entropies = [pyramid_entropy(image, a, depth) for a in a_values]
        plt.plot(a_values, entropies, label=f'Depth {depth}')
    plt.xlabel('a')
    plt.ylabel('Entropy')
    plt.title(f'Entropy vs a for Different Depths ({image_name})')
    plt.legend(title='Depth')
    plt.show()

a_values = [0.2, 0.4, 0.6, 0.8]
depths = range(2, 7)

plot_entropy_vs_a(lena_gray, depths, a_values, "Lena")

plot_entropy_vs_a_and_depth(lena_gray, a_values, depths, "Lena")
```

#### **ε) Εντοπίστε το βέλτιστο ‘a’, ως προς την εντροπία σε κάθε επίπεδο της πυραμίδας Laplacian, για κάθε εικόνα**

```python
def find_optimal_a(image, depths, a_values):
    optimal_a_values = []
    for depth in depths:
        entropy_values = [pyramid_entropy(image, a, depth) for a in a_values]
        optimal_a_index = np.argmin(entropy_values)
        optimal_a = a_values[optimal_a_index]
        optimal_a_values.append(optimal_a)
    return optimal_a_values

optimal_a_lena = find_optimal_a(lena_gray, depths, a_values)
print("Optimal 'a' values for Lena image at each level:", optimal_a_lena)

optimal_a_camera = find_optimal_a(camera_img, depths, a_values)
print("Optimal 'a' values for Camera image at each level:", optimal_a_camera)
```
